```python
import pandas as pd
from pathlib import Path

csv_path = Path("viral_videos.csv")
if not csv_path.exists():
    csv_path = Path("aula11") / csv_path
df = pd.read_csv(csv_path)
print(df.head())
```

```python
df.info()
print(f"\nTotal de colunas: {df.shape[1]}")
print(f"Total de valores nulos: {df.isna().sum().sum()}")
```

```python
print(df.describe())
print(f"\nM?dia geral de visualiza??es: {df['views'].mean():,.2f}")
```

```python
df.rename(columns={
    "category": "categoria",
    "views": "visualizacoes",
    "likes": "curtidas",
}, inplace=True)

print(df.columns.tolist())
```

```python
df_sucesso = df[df["visualizacoes"] > 500_000]
print(f"V?deos que sobraram: {len(df_sucesso)}")
print(df_sucesso.head())
```

```python
df_tech = df_sucesso[df_sucesso["categoria"] == "Tech"]
print(f"V?deos Tech de sucesso: {len(df_tech)}")
print(df_tech.head())
```

```python
media_por_thumbnail = df.groupby("thumbnail_face")["visualizacoes"].mean()
print(media_por_thumbnail)

if media_por_thumbnail.loc[1] > media_por_thumbnail.loc[0]:
    print("Veredito: v?deos com rosto t?m, em m?dia, mais visualiza??es.")
else:
    print("Veredito: v?deos sem rosto t?m, em m?dia, mais visualiza??es.")
```

```python
curtidas_por_dia = df.groupby("upload_day")["curtidas"].sum().sort_values(ascending=False)
print(curtidas_por_dia)
print(f"\nDia com maior engajamento acumulado: {curtidas_por_dia.idxmax()} ({curtidas_por_dia.max():,} curtidas).")
```
